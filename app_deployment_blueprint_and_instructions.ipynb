```python
# Simple TFR map for 1993 only
def create_simple_tfr_map_1993():
    """Create a simple TFR map for the year 1993"""
    
    # Filter for 1993 data
    test_year = 1993
    tfr_1993 = tfr_merged[tfr_merged['year'] == test_year].copy()
    
    print(f"TFR data for {test_year}:")
    print(f"Data shape: {tfr_1993.shape}")
    print(f"Number of locations: {len(tfr_1993)}")
    print(f"TFR range: {tfr_1993['tfr'].min():.3f} to {tfr_1993['tfr'].max():.3f}")
    print(f"Average TFR: {tfr_1993['tfr'].mean():.3f}")
    
    if tfr_1993.empty:
        print(f"No data available for year {test_year}")
        return None
    
    # Convert to web mercator if needed
    tfr_gdf_web = tfr_1993.to_crs('EPSG:4326') if tfr_1993.crs != 'EPSG:4326' else tfr_1993
    
    # Create the map
    fig = go.Figure()
    
    # Get TFR range for colour scaling
    tfr_min = tfr_1993['tfr'].min()
    tfr_max = tfr_1993['tfr'].max()
    
    # Create the choropleth trace
    fig.add_trace(
        go.Choroplethmapbox(
            geojson=tfr_gdf_web.__geo_interface__,
            locations=tfr_gdf_web.index,
            z=tfr_gdf_web['tfr'],
            colorscale='RdYlBu_r',  # Red-Yellow-Blue reversed (red = high fertility)
            zmin=tfr_min,
            zmax=tfr_max,
            text=tfr_gdf_web['gss_name'],
            hovertemplate='<b>%{text}</b><br>' + 
                         f'TFR ({test_year}): %{{z:.3f}}<br>' +
                         'UK Replacement Level: 2.1<extra></extra>',
            marker_opacity=0.8,
            marker_line_width=0.5,
            marker_line_color='white',
            colorbar=dict(
                title=f"Total Fertility Rate ({test_year})", 
                title_side="right",
                len=0.8,
                thickness=15
            )
        )
    )
    
    # Update layout
    fig.update_layout(
        title=dict(
            text=f'Total Fertility Rate Across UK Local Authority Districts ({test_year})',
            x=0.5,
            font=dict(size=16, color='#2E8B57')
        ),
        mapbox=dict(
            style='carto-positron',
            center={'lat': 54.5, 'lon': -2.5},
            zoom=5.2
        ),
        height=700,
        width=1100,
        font=dict(family="Arial", size=12)
    )
    
    return fig

# Create and display the simple TFR map for 1993
print("Creating simple TFR map for 1993...")
fig_tfr_1993 = create_simple_tfr_map_1993()

if fig_tfr_1993:
    fig_tfr_1993.show()
    print("✅ TFR map for 1993 created successfully!")
    print("🎯 Red areas = higher fertility rates, Blue areas = lower fertility rates")
else:
    print("❌ Failed to create TFR map for 1993")
```

```python
import dash
import dash_bootstrap_components as dbc
from dash import dcc, html, Input, Output
import plotly.express as px
import plotly.graph_objects as go

# --- Data Preparation ---
print("Preparing fertility rate data for dashboard...")

# Prepare TFR data
tfr_clean = tfr_merged[['LAD23CD', 'LAD23NM', 'geometry', 'year', 'tfr']].copy()
if tfr_clean.crs is None:
    tfr_clean = tfr_clean.set_crs("EPSG:27700", allow_override=True)
tfr_clean = tfr_clean.to_crs(epsg=4326)
tfr_clean['geometry'] = tfr_clean['geometry'].simplify(tolerance=0.008, preserve_topology=True)
tfr_clean = tfr_clean[tfr_clean.is_valid].reset_index(drop=True)

# Prepare ASFR data  
asfr_clean = asfr_merged[['LAD23CD','LAD23NM','geometry','age','year','fertility_rate']].copy()
if asfr_clean.crs is None:
    asfr_clean = asfr_clean.set_crs("EPSG:27700", allow_override=True)
asfr_clean = asfr_clean.to_crs(epsg=4326)
asfr_clean['geometry'] = asfr_clean['geometry'].simplify(tolerance=0.008, preserve_topology=True)
asfr_clean = asfr_clean[asfr_clean.is_valid].reset_index(drop=True)

# Get available options
years = sorted(tfr_clean['year'].unique())
ages = sorted(asfr_clean['age'].unique())
lads = sorted(tfr_clean['LAD23NM'].unique())

print(f"✅ Data prepared: {len(years)} years, {len(ages)} ages, {len(lads)} local authorities")

# --- Initialise Dash App ---
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])
app.title = "UK Fertility Rate Dashboard"

# --- App Layout ---
app.layout = dbc.Container([
    # Header
    dbc.Row([
        dbc.Col([
            html.H1("🇬🇧 UK Fertility Rate Dashboard", 
                   className="text-center mb-4",
                   style={'color': '#2E8B57', 'fontWeight': 'bold'}),
            html.Hr(style={'borderColor': '#FF6B6B', 'borderWidth': '3px'})
        ])
    ]),
    
    # Control Panel
    dbc.Card([
        dbc.CardBody([
            html.H4("📊 Dashboard Controls", className="mb-3", style={'color': '#2E8B57'}),
            
            dbc.Row([
                # Year Selection
                dbc.Col([
                    html.Label("📅 Select Year:", className="fw-bold"),
                    dcc.Dropdown(
                        id='year-dropdown',
                        options=[{'label': f"{year}", 'value': year} for year in years],
                        value=years[-1],  # Start with most recent year
                        clearable=False,
                        style={'marginBottom': '10px'}
                    )
                ], width=3),
                
                # Age Selection (for ASFR)
                dbc.Col([
                    html.Label("👶 Select Age (ASFR):", className="fw-bold"),
                    dcc.Dropdown(
                        id='age-dropdown',
                        options=[{'label': f"Age {age}", 'value': age} for age in ages],
                        value=30,  # Default to age 30
                        clearable=False,
                        style={'marginBottom': '10px'}
                    )
                ], width=3),
                
                # LAD Selection (for line graphs)
                dbc.Col([
                    html.Label("🏘️ Select Local Authority:", className="fw-bold"),
                    dcc.Dropdown(
                        id='lad-dropdown',
                        options=[{'label': lad, 'value': lad} for lad in lads[:50]], # Show first 50 for performance
                        value="Manchester",  # Default LAD
                        clearable=False,
                        style={'marginBottom': '10px'}
                    )
                ], width=6)
            ])
        ])
    ], className="mb-4", style={'backgroundColor': '#F8F9FA'}),
    
    # Maps Row
    dbc.Row([
        # TFR Map
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("🗺️ Total Fertility Rate (TFR) Map", className="text-center mb-0")),
                dbc.CardBody([
                    dcc.Graph(id='tfr-map', style={'height': '500px'})
                ])
            ])
        ], width=6),
        
        # ASFR Map
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("🗺️ Age-Specific Fertility Rate (ASFR) Map", className="text-center mb-0")),
                dbc.CardBody([
                    dcc.Graph(id='asfr-map', style={'height': '500px'})
                ])
            ])
        ], width=6)
    ], className="mb-4"),
    
    # Line Graphs Row
    dbc.Row([
        # TFR Trend
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("📈 TFR Trend Over Time", className="text-center mb-0")),
                dbc.CardBody([
                    dcc.Graph(id='tfr-line', style={'height': '400px'})
                ])
            ])
        ], width=6),
        
        # ASFR Trend
        dbc.Col([
            dbc.Card([
                dbc.CardHeader(html.H5("📈 ASFR Trend Over Time", className="text-center mb-0")),
                dbc.CardBody([
                    dcc.Graph(id='asfr-line', style={'height': '400px'})
                ])
            ])
        ], width=6)
    ], className="mb-4"),
    
    # Footer
    dbc.Row([
        dbc.Col([
            html.Hr(),
            html.P("💡 This dashboard visualises UK fertility rates using ONS data. TFR shows total fertility rate, ASFR shows age-specific fertility rate.",
                   className="text-muted text-center"),
            html.P("🔄 Use the controls above to explore different years, ages, and local authorities.",
                   className="text-muted text-center")
        ])
    ])
    
], fluid=True, style={'padding': '20px'})

# --- Metric Cards Callback ---
@app.callback(
    [Output('tfr-title', 'children'),
     Output('comparison-title', 'children'),
     Output('current-tfr', 'children'),
     Output('tfr-comparison', 'children'),
     Output('replacement-status', 'children')],
    [Input('year-dropdown', 'value'),
     Input('lad-dropdown', 'value')]
)
def update_metrics(selected_year, selected_lad):
    if not selected_year or not selected_lad:
        return (
            "Total Fertility Rate",
            "vs UK Average", 
            "--", 
            "--", 
            "--"
        )
    
    # Get LAD TFR for selected year and area
    lad_tfr = tfr_clean[
        (tfr_clean['year'] == selected_year) & 
        (tfr_clean['LAD23NM'] == selected_lad)
    ]
    
    # Get UK average for selected year
    uk_avg = tfr_clean[tfr_clean['year'] == selected_year]['tfr'].mean()
    
    if lad_tfr.empty:
        current_tfr_val = "--"
        comparison_val = "--"
        replacement_val = "--"
    else:
        current_tfr_val = f"{lad_tfr.iloc[0]['tfr']:.2f}"
        
        # Calculate difference from UK average
        diff = lad_tfr.iloc[0]['tfr'] - uk_avg
        if diff > 0:
            comparison_val = f"+{diff:.2f}"
        else:
            comparison_val = f"{diff:.2f}"
        
        # Replacement level status
        tfr_val = lad_tfr.iloc[0]['tfr']
        if tfr_val >= 2.1:
            replacement_val = "Above replacement"
        else:
            replacement_val = f"{2.1 - tfr_val:.2f} below"
    
    # Dynamic titles
    tfr_title = f"Total Fertility Rate for {selected_lad} in {selected_year}"
    comparison_title = f"{selected_lad} vs UK Average"
    
    return tfr_title, comparison_title, current_tfr_val, comparison_val, replacement_val

# --- TFR Map Callback ---
@app.callback(
    Output('tfr-map', 'figure'),
    Input('year-dropdown', 'value')
)
def update_tfr_map(selected_year):
    # Filter data
    filtered_tfr = tfr_clean[tfr_clean['year'] == selected_year]
    
    if filtered_tfr.empty:
        return go.Figure().add_annotation(text="No data available", showarrow=False)
    
    # Create choropleth map
    fig = px.choropleth_mapbox(
        filtered_tfr,
        geojson=filtered_tfr.__geo_interface__,
        locations='LAD23CD',
        featureidkey='properties.LAD23CD',
        color='tfr',
        hover_name='LAD23NM',
        hover_data={'tfr': ':.3f', 'year': True, 'LAD23CD': False},
        color_continuous_scale='RdYlBu_r',  # Red=high, Blue=low fertility
        mapbox_style='carto-positron',
        zoom=5.2,
        center={"lat": 54.5, "lon": -2.5},
        opacity=0.8,
        title=f"TFR in {selected_year}"
    )
    
    fig.update_layout(
        margin={"r":0,"t":40,"l":0,"b":0},
        title_font_size=14,
        coloraxis_colorbar=dict(
            title="TFR",
            title_font_size=12
        )
    )
    return fig

# --- ASFR Map Callback ---
@app.callback(
    Output('asfr-map', 'figure'),
    Input('year-dropdown', 'value'),
    Input('age-dropdown', 'value')
)
def update_asfr_map(selected_year, selected_age):
    # Filter data
    filtered_asfr = asfr_clean[
        (asfr_clean['year'] == selected_year) &
        (asfr_clean['age'] == selected_age)
    ]
    
    if filtered_asfr.empty:
        return go.Figure().add_annotation(text="No data available", showarrow=False)
    
    # Create choropleth map
    fig = px.choropleth_mapbox(
        filtered_asfr,
        geojson=filtered_asfr.__geo_interface__,
        locations='LAD23CD',
        featureidkey='properties.LAD23CD',
        color='fertility_rate',
        hover_name='LAD23NM',
        hover_data={'fertility_rate': ':.4f', 'year': True, 'age': True, 'LAD23CD': False},
        color_continuous_scale='Plasma',
        mapbox_style='carto-positron',
        zoom=5.2,
        center={"lat": 54.5, "lon": -2.5},
        opacity=0.8,
        title=f"ASFR Age {selected_age} in {selected_year}"
    )
    
    fig.update_layout(
        margin={"r":0,"t":40,"l":0,"b":0},
        title_font_size=14,
        coloraxis_colorbar=dict(
            title="ASFR",
            title_font_size=12
        )
    )
    return fig

# --- TFR Line Graph Callback ---
@app.callback(
    Output('tfr-line', 'figure'),
    Input('lad-dropdown', 'value')
)
def update_tfr_line(selected_lad):
    if not selected_lad:
        return go.Figure().add_annotation(text="Please select a Local Authority", showarrow=False)
    
    # Filter data
    filtered_tfr = tfr_clean[tfr_clean['LAD23NM'] == selected_lad]
    
    if filtered_tfr.empty:
        return go.Figure().add_annotation(text=f"No TFR data for {selected_lad}", showarrow=False)
    
    # Create line graph
    fig = px.line(
        filtered_tfr,
        x='year',
        y='tfr',
        title=f"TFR Trend: {selected_lad}",
        markers=True,
        line_shape='spline'
    )
    
    # Add replacement level reference line
    fig.add_hline(
        y=2.1, 
        line_dash="dash", 
        line_color="red",
        annotation_text="UK Replacement Level (2.1)",
        annotation_position="bottom right"
    )
    
    fig.update_layout(
        xaxis_title="Year",
        yaxis_title="Total Fertility Rate",
        title_font_size=14,
        height=380
    )
    
    return fig

# --- ASFR Line Graph Callback ---
@app.callback(
    Output('asfr-line', 'figure'),
    Input('lad-dropdown', 'value'),
    Input('age-dropdown', 'value')
)
def update_asfr_line(selected_lad, selected_age):
    if not selected_lad:
        return go.Figure().add_annotation(text="Please select a Local Authority", showarrow=False)
    
    # Filter data
    filtered_asfr = asfr_clean[
        (asfr_clean['LAD23NM'] == selected_lad) &
        (asfr_clean['age'] == selected_age)
    ]
    
    if filtered_asfr.empty:
        return go.Figure().add_annotation(text=f"No ASFR data for {selected_lad}, Age {selected_age}", showarrow=False)
    
    # Create line graph
    fig = px.line(
        filtered_asfr,
        x='year',
        y='fertility_rate',
        title=f"ASFR Trend: {selected_lad} (Age {selected_age})",
        markers=True,
        line_shape='spline',
        color_discrete_sequence=['#FF6B6B']
    )
    
    fig.update_layout(
        xaxis_title="Year",
        yaxis_title="Age-Specific Fertility Rate",
        title_font_size=14,
        height=380
    )
    
    return fig

# --- Run App ---
if __name__ == '__main__':
    print("🚀 Starting Fertility Rate Dashboard...")
    print("📍 Access your dashboard at: http://127.0.0.1:8050")
    print("💡 Use Ctrl+C to stop the server")
    app.run(debug=True, port=8090)
```
